# QM9s-RPFR quickstart

Load the 1% subsample, inspect a molecule, read β at 300 K, and compute dataset statistics.

```python
from pathlib import Path

from rpfr_visualizer import QM9sDataset

DATA_PATH = Path("../data/raw/qm9s_v2.2_1pt.h5")
ds = QM9sDataset(DATA_PATH)
print(f"Loaded {len(ds)} molecules")
```

```python
mol = ds.get_molecule(ds.molecule_ids[0])
print(f"SMILES: {mol.smiles}")
print(f"Atoms: {mol.n_atoms}, heavy: {mol.heavy_atom_count}")
print(f"QC pass: {mol.qc_pass}")
print(f"Site 0 beta at 300 K: {mol.sites[0].beta_at(300.0):.4f}")
```

```python
report = ds.validate()
report.summary()
```

```python
stats = ds.stats(temperature=300.0)
stats["element_counts"], stats["qc_pass_rate"]
```
